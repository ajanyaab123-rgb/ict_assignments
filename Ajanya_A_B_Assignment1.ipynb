# ⃣4 Maximum Finder
Task:
● Create a 4×4 random matrix.
● Find the maximum value.
● Find its position (row, column).
Hint:
First find the index in flattened form, then convert it back to row and column

```python
arr=np.arange(1,17)
arr1=arr.reshape(4,4)
arr1
```

```python
max_value=np.max(arr1)
print('Maximum value:',max_value)
```

```python
max_index=np.argmax(arr1)
max_position=np.unravel_index(max_index,arr1.shape)
print('Position of maximum value:',max_position)
```

8️⃣Filter Rows Based on Condition
Task:
● Create a 5×3 random matrix (values between 0–20).
● Print only rows where row sum > 30.

```python
X = np.random.randint(0, 21, (5, 3))

print(X)

print(X[X.sum(axis=1) > 30])
```

🔟 Simple Normalization
Task:
● Create a 3×3 random matrix.
● Divide each row by its row sum.

```python
X = np.random.rand(3, 3)

result = X / X.sum(axis=1, keepdims=True)

print(result)
```
